```python
import json
import pandas as pd
from glob import glob
import os

data_dict = {}
for item in os.listdir(os.path.expanduser("~/Downloads/math/")):
    evaluator = item.split("_")[-1]
    data_dict[evaluator] = {}
    print(item)
    if "evaluator" in item:
        for file in os.listdir(os.path.join(os.path.expanduser("~/Downloads/math/"), item)):
            print(file)
            data = []
            with open(os.path.join(os.path.expanduser("~/Downloads/math/"), item, file), "r") as f:
                for line in f.readlines():
                    data.append(json.loads(line))
            data_dict[evaluator][file] = data
```

```python
llama_preferences = {}
for judge in data_dict.keys():
    for combination in data_dict[judge].keys():
        if "average" in combination or "merge" in combination:
            continue
        m1, m2 = combination.replace(".jsonl", "").split("_")
        print(judge, m1, m2)
        if judge in [m1, m2]:
            print("Self evaluation")
        for item in data_dict[judge][combination][:1]:
            print(item['response1'])
    break
```
